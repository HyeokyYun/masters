### GridSearchCV를 이용한 최적 하이퍼파라미터 찾기

```python
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
import numpy as np

# train_test_split 함수에 stratify 매개변수를 y로 설정
X_train, X_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.3, random_state=11, stratify=df_y)


# 레이블 인코딩 및 원-핫 인코딩
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)
y_train_encoded = to_categorical(y_train_encoded, num_classes=4)
y_test_encoded = to_categorical(y_test_encoded, num_classes=4)

# 모델 생성 함수
def create_model(activation, optimizer):
    model = Sequential()
    model.add(Dense(128, activation=activation, input_shape=(X_train.shape[1],)))
    model.add(Dense(64, activation=activation))
    model.add(Dense(32, activation=activation))
    model.add(Dense(4, activation='softmax'))  # 클래스 개수에 맞게 조정
    model.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy', f1_score])
    return model

# KerasClassifier 설정
model = KerasClassifier(model=create_model, verbose=0)

# 하이퍼파라미터 그리드 설정
param_grid = {
    'model__activation': ['relu', 'tanh'],
    'model__optimizer': ['adam', 'rmsprop', 'sgd'],
    'batch_size': [24, 32],
    'epochs': [10, 20]
}

# 그리드 탐색
grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=5)
grid_result = grid.fit(X_train, y_train_encoded)

# 결과 출력
print("Best: %f using %s" % (grid_result.best_score_, grid_result.best_params_))

means = grid_result.cv_results_['mean_test_score']
stds = grid_result.cv_results_['std_test_score']
params = grid_result.cv_results_['params']

for mean, stdev, param in zip(means, stds, params):
    print("%f (%f) with: %r" % (mean, stdev, param))
```

## 108주 Time-Series 샘플

```python
df = pd.read_csv('D:/credit/training/sample/ML_training_sample_2y.csv')
print(df.shape)
df
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# 'feature1' 열에 대해 Min-Max Scaling 적용
df['total_purchase_invoice'] = scaler.fit_transform(df[['total_purchase_invoice']])
df['total_purchase_card'] = scaler.fit_transform(df[['total_purchase_card']])

df
```

```python
# 시간 변수 선택
X = df.iloc[:, 14:-2]
y = df['new_label']

# time-invarying 변수 선택
static_feature = df[['total_purchase_invoice', 'total_purchase_card',
                'sales_05', 'sales_611', 'sales_1823', 'sales_1217', 'weekend_sales', 'delivery_ratio',
               'new_customer_ratio', 'delivery_link', 'franchise1', 'open_after', 'category']]
```

```python
# 업종변수 더미변수화

category_dummy = pd.get_dummies(static_feature['category'], prefix='category')
static_feature1 = pd.concat([static_feature, category_dummy], axis=1)
static_feature1['category_cafe'] = static_feature1['category_cafe'].astype(int)
static_feature1['category_hof'] = static_feature1['category_hof'].astype(int)
static_feature1['category_normal'] = static_feature1['category_normal'].astype(int)

static_feature1.drop('category', axis=1, inplace=True)
```

```python
# Long format으로 전환

expanded_static_feature = np.repeat(static_feature1.values[:, np.newaxis, :], 109, axis=1)
combined_data = np.concatenate((X.values[:, :, np.newaxis], expanded_static_feature), axis=2)
```

```python
X_reshaped = combined_data
X_reshaped
```

```python
#데이터 형태 확인

X_reshaped.shape
```

```python
# 테스트셋, 트레인셋 데이터 나누기

X_train, X_test, y_train, y_test = train_test_split(X_reshaped, y, test_size=0.3, random_state=42, stratify = y)
```

```python
# 클래스 개수 확인

y_train.value_counts()
```

```python
# 타겟변수에 대한 원-핫 인코딩
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

y_train_encoded = to_categorical(y_train_encoded, num_classes=6)
y_test_encoded = to_categorical(y_test_encoded, num_classes=6)
```

### 1D CNN 모델

```python
from tensorflow.keras.layers import ConvLSTM2D, Flatten
from sklearn.metrics import f1_score

# 합성곱 LSTM 모델 구성
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense

# label별 가중치 설정 (데이터가 많을수록 가중치 낮음)
weight_for_0 = ( 1/ 1004)*(1004+6269+723+81+879+1086)/6
weight_for_1 = ( 1/ 6269)*(1004+6269+723+81+879+1086)/6
weight_for_2 = (1 / 723)*(1004+6269+723+81+879+1086)/6
weight_for_3 = (1 / 81)*(1004+6269+723+81+879+1086)/6
weight_for_4 = (1 / 879)*(1004+6269+723+81+879+1086)/6
weight_for_5 = (1 / 1086)*(1004+6269+723+81+879+1086)/6

class_weights_dic = {0: weight_for_0, 1: weight_for_1, 2:weight_for_2, 
                    3: weight_for_3, 4:weight_for_4, 5: weight_for_5}
# 모델 구성
model = Sequential()

# 1D 컨볼루션 레이어 추가
# 예시 매개변수: 64개의 필터, 커널 크기 3
model.add(Conv1D(filters=64, kernel_size=3, activation='relu', input_shape=(109, 16)))

# 1D 맥스 풀링 레이어 추가
model.add(MaxPooling1D(pool_size=3))

# Flatten 레이어 추가
model.add(Flatten())

# 완전 연결 레이어 추가
model.add(Dense(32, activation='relu'))


# 출력 레이어 추가
model.add(Dense(6, activation='softmax')) 

# 모델 컴파일
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# 모델 학습
model.fit(X_train, y_train_encoded, epochs=10, batch_size=24, validation_split=0.2, class_weight = class_weights_dic)

# 모델 평가
y_pred = model.predict(X_test)

# 원-핫 인코딩된 예측 결과를 정수 레이블로 변환
y_pred_classes = np.argmax(y_pred, axis=1)

# 실제 레이블이 원-핫 인코딩 형태인 경우, 정수 레이블로 변환
y_test_classes = np.argmax(y_test_encoded, axis=1)

# 성능 메트릭 계산
print("Accuracy:", accuracy_score(y_test_classes, y_pred_classes))
print("f1:", f1_score(y_test_classes, y_pred_classes, average = 'weighted'))
```
