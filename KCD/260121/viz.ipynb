```python
import pandas as pd
import numpy as np

print("=" * 80)
print("전체 대비 비율 높은 순위 계산 (하위 그룹 제외)")
print("=" * 80)

# ============================================================================
# 1. 데이터 로드
# ============================================================================
print("\n[1] 데이터 로드 중...")

sigungu_ratio = pd.read_csv('cluster_sigungu_relative_ratio.csv')
dong_ratio = pd.read_csv('cluster_dong_relative_ratio.csv')
depth_1_ratio = pd.read_csv('cluster_depth_1_relative_ratio.csv')
depth_2_ratio = pd.read_csv('cluster_depth_2_relative_ratio.csv')
depth_3_ratio = pd.read_csv('cluster_depth_3_relative_ratio.csv')

print("데이터 로드 완료")

# ============================================================================
# 2. 하위 그룹 필터링 및 상위 순위 계산
# ============================================================================

def filter_and_rank(df, category_name, min_cluster_count=5, min_total_count=10):
    """
    하위 그룹을 필터링하고 비율 높은 순으로 정렬
    """
    # 하위 그룹 필터링
    filtered = df[
        (df['cluster_count'] >= min_cluster_count) & 
        (df['total_count'] >= min_total_count)
    ].copy()
    
    # 비율 높은 순으로 정렬
    results = []
    for cluster_id in range(9):
        cluster_data = filtered[filtered['cluster'] == cluster_id].copy()
        cluster_data = cluster_data.sort_values('ratio', ascending=False)
        results.append(cluster_data)
    
    result_df = pd.concat(results, ignore_index=True)
    return result_df

# sigungu: 최소 클러스터 내 5개, 전체 20개 이상
print("\n[2] sigungu 상위 순위 계산 중...")
sigungu_ranked = filter_and_rank(sigungu_ratio, 'sigungu', min_cluster_count=5, min_total_count=30)
print(f"필터링 후: {len(sigungu_ranked)}개 레코드 (원본: {len(sigungu_ratio)}개)")

# dong: 최소 클러스터 내 3개, 전체 10개 이상 (더 엄격하게)
print("\n[3] dong 상위 순위 계산 중...")
dong_ranked = filter_and_rank(dong_ratio, 'dong', min_cluster_count=3, min_total_count=30)
print(f"필터링 후: {len(dong_ranked)}개 레코드 (원본: {len(dong_ratio)}개)")

# depth_1: 최소 클러스터 내 10개, 전체 50개 이상
print("\n[4] depth_1 상위 순위 계산 중...")
depth_1_ranked = filter_and_rank(depth_1_ratio, 'depth_1', min_cluster_count=10, min_total_count=50)
print(f"필터링 후: {len(depth_1_ranked)}개 레코드 (원본: {len(depth_1_ratio)}개)")

# depth_2: 최소 클러스터 내 5개, 전체 20개 이상
print("\n[5] depth_2 상위 순위 계산 중...")
depth_2_ranked = filter_and_rank(depth_2_ratio, 'depth_2', min_cluster_count=5, min_total_count=30)
print(f"필터링 후: {len(depth_2_ranked)}개 레코드 (원본: {len(depth_2_ratio)}개)")

# depth_3: 최소 클러스터 내 3개, 전체 10개 이상
print("\n[6] depth_3 상위 순위 계산 중...")
depth_3_ranked = filter_and_rank(depth_3_ratio, 'depth_3', min_cluster_count=3, min_total_count=30)
print(f"필터링 후: {len(depth_3_ranked)}개 레코드 (원본: {len(depth_3_ratio)}개)")

# ============================================================================
# 3. 결과 저장
# ============================================================================
print("\n[7] 결과 저장 중...")

sigungu_ranked.to_csv('cluster_sigungu_top_ratios.csv', index=False, encoding='utf-8-sig')
print("저장 완료: cluster_sigungu_top_ratios.csv")

dong_ranked.to_csv('cluster_dong_top_ratios.csv', index=False, encoding='utf-8-sig')
print("저장 완료: cluster_dong_top_ratios.csv")

depth_1_ranked.to_csv('cluster_depth_1_top_ratios.csv', index=False, encoding='utf-8-sig')
print("저장 완료: cluster_depth_1_top_ratios.csv")

depth_2_ranked.to_csv('cluster_depth_2_top_ratios.csv', index=False, encoding='utf-8-sig')
print("저장 완료: cluster_depth_2_top_ratios.csv")

depth_3_ranked.to_csv('cluster_depth_3_top_ratios.csv', index=False, encoding='utf-8-sig')
print("저장 완료: cluster_depth_3_top_ratios.csv")

# ============================================================================
# 4. 상위 순위 출력 (각 클러스터별 상위 5개)
# ============================================================================
print("\n[8] 클러스터별 상위 5개 출력:")

print("\n=== sigungu ===")
for cluster_id in range(9):
    cluster_data = sigungu_ranked[sigungu_ranked['cluster'] == cluster_id].head(10)
    if len(cluster_data) > 0:
        print(f"\nCluster {cluster_id}:")
        for _, row in cluster_data.iterrows():
            print(f"  {row['sigungu']}: {row['ratio']*100:.2f}% ({int(row['cluster_count'])}/{int(row['total_count'])})")

print("\n=== dong ===")
for cluster_id in range(9):
    cluster_data = dong_ranked[dong_ranked['cluster'] == cluster_id].head(10)
    if len(cluster_data) > 0:
        print(f"\nCluster {cluster_id}:")
        for _, row in cluster_data.iterrows():
            print(f"  {row['dong']}: {row['ratio']*100:.2f}% ({int(row['cluster_count'])}/{int(row['total_count'])})")

print("\n=== depth_2 ===")
for cluster_id in range(9):
    cluster_data = depth_2_ranked[depth_2_ranked['cluster'] == cluster_id].head(10)
    if len(cluster_data) > 0:
        print(f"\nCluster {cluster_id}:")
        for _, row in cluster_data.iterrows():
            print(f"  {row['depth_2']}: {row['ratio']*100:.2f}% ({int(row['cluster_count'])}/{int(row['total_count'])})")

print("\n=== depth_3 ===")
for cluster_id in range(9):
    cluster_data = depth_3_ranked[depth_3_ranked['cluster'] == cluster_id].head(10)
    if len(cluster_data) > 0:
        print(f"\nCluster {cluster_id}:")
        for _, row in cluster_data.iterrows():
            print(f"  {row['depth_3']}: {row['ratio']*100:.2f}% ({int(row['cluster_count'])}/{int(row['total_count'])})")

print("\n" + "=" * 80)
print("작업 완료!")
print("=" * 80)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정
try:
    plt.rcParams['font.family'] = 'AppleGothic'
except:
    pass
plt.rcParams['axes.unicode_minus'] = False

# ============================================================================
# 1. 데이터 로드 및 필터링
# ============================================================================
print("데이터 로드 및 필터링 중...")

sigungu_ratio = pd.read_csv('cluster_sigungu_relative_ratio.csv')
dong_ratio = pd.read_csv('cluster_dong_relative_ratio.csv')
depth_1_ratio = pd.read_csv('cluster_depth_1_relative_ratio.csv')
depth_2_ratio = pd.read_csv('cluster_depth_2_relative_ratio.csv')
depth_3_ratio = pd.read_csv('cluster_depth_3_relative_ratio.csv')

def filter_and_rank(df, min_cluster_count=5, min_total_count=10):
    """하위 그룹 필터링 및 비율 높은 순 정렬"""
    filtered = df[
        (df['cluster_count'] >= min_cluster_count) & 
        (df['total_count'] >= min_total_count)
    ].copy()
    
    results = []
    for cluster_id in range(9):
        cluster_data = filtered[filtered['cluster'] == cluster_id].copy()
        cluster_data = cluster_data.sort_values('ratio', ascending=False)
        results.append(cluster_data)
    
    return pd.concat(results, ignore_index=True)

# 필터링 적용
sigungu_ranked = filter_and_rank(sigungu_ratio, min_cluster_count=5, min_total_count=20)
dong_ranked = filter_and_rank(dong_ratio, min_cluster_count=3, min_total_count=10)
depth_1_ranked = filter_and_rank(depth_1_ratio, min_cluster_count=10, min_total_count=50)
depth_2_ranked = filter_and_rank(depth_2_ratio, min_cluster_count=5, min_total_count=20)
depth_3_ranked = filter_and_rank(depth_3_ratio, min_cluster_count=3, min_total_count=10)

print("필터링 완료")

# ============================================================================
# 2. sigungu 막대그래프 (비율 기준)
# ============================================================================
print("\nsigungu 막대그래프 생성 중...")
fig, axes = plt.subplots(3, 3, figsize=(20, 18))
axes = axes.flatten()

for cluster_id in range(9):
    ax = axes[cluster_id]
    cluster_data = sigungu_ranked[sigungu_ranked['cluster'] == cluster_id].head(10).sort_values('ratio', ascending=True)
    
    if len(cluster_data) > 0:
        y_pos = np.arange(len(cluster_data))
        ratios = cluster_data['ratio'].values * 100  # 비율 (%)
        labels = cluster_data['sigungu'].values
        
        bars = ax.barh(y_pos, ratios, color='steelblue', edgecolor='black', linewidth=1)
        ax.set_yticks(y_pos)
        ax.set_yticklabels(labels, fontsize=14)
        ax.set_xlabel('전체 대비 비율 (%)', fontsize=14, fontweight='bold')
        ax.set_title(f'Cluster {cluster_id} - 상위 10개', fontsize=14, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='x')
        
        # 비율과 개수 표시
        for bar, ratio, row in zip(bars, ratios, cluster_data.itertuples()):
            width = bar.get_width()
            ax.text(width, bar.get_y() + bar.get_height()/2.,
                   f'{ratio:.1f}% ({int(row.cluster_count)}/{int(row.total_count)})',
                   ha='left', va='center', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('bar_chart_sigungu_top_ratios.png', dpi=300, bbox_inches='tight')
plt.show()
print("저장 완료: bar_chart_sigungu_top_ratios.png")

# ============================================================================
# 3. dong 막대그래프 (비율 기준)
# ============================================================================
print("\ndong 막대그래프 생성 중...")
fig, axes = plt.subplots(3, 3, figsize=(20, 18))
axes = axes.flatten()

for cluster_id in range(9):
    ax = axes[cluster_id]
    cluster_data = dong_ranked[dong_ranked['cluster'] == cluster_id].head(10).sort_values('ratio', ascending=True)
    
    if len(cluster_data) > 0:
        y_pos = np.arange(len(cluster_data))
        ratios = cluster_data['ratio'].values * 100
        labels = cluster_data['dong'].values
        
        bars = ax.barh(y_pos, ratios, color='steelblue', edgecolor='black', linewidth=1)
        ax.set_yticks(y_pos)
        ax.set_yticklabels(labels, fontsize=14)
        ax.set_xlabel('전체 대비 비율 (%)', fontsize=14, fontweight='bold')
        ax.set_title(f'Cluster {cluster_id} - 상위 10개', fontsize=14, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='x')
        
        for bar, ratio, row in zip(bars, ratios, cluster_data.itertuples()):
            width = bar.get_width()
            ax.text(width, bar.get_y() + bar.get_height()/2.,
                   f'{ratio:.1f}% ({int(row.cluster_count)}/{int(row.total_count)})',
                   ha='left', va='center', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('bar_chart_dong_top_ratios.png', dpi=300, bbox_inches='tight')
plt.show()
print("저장 완료: bar_chart_dong_top_ratios.png")

# ============================================================================
# 4. depth_1 막대그래프 (비율 기준)
# ============================================================================
print("\ndepth_1 막대그래프 생성 중...")
fig, ax = plt.subplots(figsize=(10, 6))
clusters = sorted(depth_1_ranked['cluster'].unique())
ratios = [depth_1_ranked[depth_1_ranked['cluster'] == c]['ratio'].values[0] * 100 for c in clusters]

bars = ax.bar(range(len(clusters)), ratios, color='steelblue', edgecolor='black', linewidth=1.5)
ax.set_xlabel('Cluster', fontsize=14, fontweight='bold')
ax.set_ylabel('전체 대비 비율 (%)', fontsize=14, fontweight='bold')
ax.set_title('클러스터별 depth_1 (외식업) 전체 대비 비율', fontsize=14, fontweight='bold')
ax.set_xticks(range(len(clusters)))
ax.set_xticklabels([f'Cluster {c}' for c in clusters], fontsize=14)
ax.grid(True, alpha=0.3, axis='y')

for bar, ratio, c in zip(bars, ratios, clusters):
    row = depth_1_ranked[depth_1_ranked['cluster'] == c].iloc[0]
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
           f'{ratio:.1f}% ({int(row.cluster_count)}/{int(row.total_count)})',
           ha='center', va='bottom', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('bar_chart_depth_1_top_ratios.png', dpi=300, bbox_inches='tight')
plt.show()
print("저장 완료: bar_chart_depth_1_top_ratios.png")

# ============================================================================
# 5. depth_2 막대그래프 (비율 기준)
# ============================================================================
print("\ndepth_2 막대그래프 생성 중...")
fig, axes = plt.subplots(3, 3, figsize=(20, 18))
axes = axes.flatten()

for cluster_id in range(9):
    ax = axes[cluster_id]
    cluster_data = depth_2_ranked[depth_2_ranked['cluster'] == cluster_id].head(10).sort_values('ratio', ascending=True)
    
    if len(cluster_data) > 0:
        y_pos = np.arange(len(cluster_data))
        ratios = cluster_data['ratio'].values * 100
        labels = cluster_data['depth_2'].values
        
        bars = ax.barh(y_pos, ratios, color='steelblue', edgecolor='black', linewidth=1)
        ax.set_yticks(y_pos)
        ax.set_yticklabels(labels, fontsize=14)
        ax.set_xlabel('전체 대비 비율 (%)', fontsize=14, fontweight='bold')
        ax.set_title(f'Cluster {cluster_id} - 상위 10개', fontsize=14, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='x')
        
        for bar, ratio, row in zip(bars, ratios, cluster_data.itertuples()):
            width = bar.get_width()
            ax.text(width, bar.get_y() + bar.get_height()/2.,
                   f'{ratio:.1f}% ({int(row.cluster_count)}/{int(row.total_count)})',
                   ha='left', va='center', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('bar_chart_depth_2_top_ratios.png', dpi=300, bbox_inches='tight')
plt.show()
print("저장 완료: bar_chart_depth_2_top_ratios.png")

# ============================================================================
# 6. depth_3 막대그래프 (비율 기준)
# ============================================================================
print("\ndepth_3 막대그래프 생성 중...")
fig, axes = plt.subplots(3, 3, figsize=(20, 18))
axes = axes.flatten()

for cluster_id in range(9):
    ax = axes[cluster_id]
    cluster_data = depth_3_ranked[depth_3_ranked['cluster'] == cluster_id].head(10).sort_values('ratio', ascending=True)
    
    if len(cluster_data) > 0:
        y_pos = np.arange(len(cluster_data))
        ratios = cluster_data['ratio'].values * 100
        labels = cluster_data['depth_3'].values
        
        bars = ax.barh(y_pos, ratios, color='steelblue', edgecolor='black', linewidth=1)
        ax.set_yticks(y_pos)
        ax.set_yticklabels(labels, fontsize=14)
        ax.set_xlabel('전체 대비 비율 (%)', fontsize=14, fontweight='bold')
        ax.set_title(f'Cluster {cluster_id} - 상위 10개', fontsize=14, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='x')
        
        for bar, ratio, row in zip(bars, ratios, cluster_data.itertuples()):
            width = bar.get_width()
            ax.text(width, bar.get_y() + bar.get_height()/2.,
                   f'{ratio:.1f}% ({int(row.cluster_count)}/{int(row.total_count)})',
                   ha='left', va='center', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('bar_chart_depth_3_top_ratios.png', dpi=300, bbox_inches='tight')
plt.show()
print("저장 완료: bar_chart_depth_3_top_ratios.png")

print("\n모든 막대그래프 생성 완료!")
```
