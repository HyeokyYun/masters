```python
# GPU 기반 DTW K-Means 클러스터링
# 필요한 패키지 설치: pip install torch pytorch-softdtw-cuda

import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score

# GPU 사용 가능 여부 확인
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")

# Soft-DTW 임포트 (현재 서버에 있는 로컬 구현 사용)
from soft_dtw_cuda import SoftDTW
print("Using soft_dtw_cuda (local)")

# ============================================
# 데이터 로드 및 전처리 (기존 코드와 동일)
# ============================================
df = pd.read_csv("./sample_2019_w162.csv")
df = df[~df['day_after1'].isin([0])]

# id별 시리즈 생성
grouped = df.groupby('public_id')
series_dict = {pid: group.set_index('day_after1')['sales_ratio'] for pid, group in grouped}
mySeries = {key: series.sort_index() for key, series in series_dict.items()}
namesofMyseries = list(mySeries.keys())

# 최소 길이 필터링
mySeries1 = [series for series in mySeries.values() if len(series) >= 95]

# 최대 길이 찾기
max_len = max(len(series) for series in mySeries1)

# 패딩 및 보간
for i in range(len(mySeries1)):
    if len(mySeries1[i]) != max_len:
        longest_series = max(mySeries1, key=len)
        mySeries1[i] = mySeries1[i].reindex(longest_series.index)
        mySeries1[i].interpolate(limit_direction="both", inplace=True)

# MinMax 스케일링
for i in range(len(mySeries1)):
    scaler = MinMaxScaler()
    data_reshaped = np.array(mySeries1[i]).reshape(-1, 1)
    mySeries1[i] = scaler.fit_transform(data_reshaped).reshape(-1)

# ============================================
# GPU 기반 DTW K-Means 클러스터링
# ============================================

# 데이터를 PyTorch 텐서로 변환
# shape: (N, L, 1) where N=시계열 개수, L=시계열 길이
#X_np = np.array([s.values for s in mySeries1])  # (N, L)
X_np = np.array([np.asarray(s) for s in mySeries1], dtype=np.float32)  # (N, L)
X_np = X_np[:, :, np.newaxis]  # (N, L, 1)
#X = torch.FloatTensor(X_np).to(device)
X = torch.from_numpy(X_np).to(device)

print(f"Data shape: {X.shape}")
print(f"Number of time series: {X.shape[0]}")
print(f"Time series length: {X.shape[1]}")

# 클러스터링 파라미터
K = 9
max_iters = 50
tol = 1e-4
gamma = 0.1  # Soft-DTW smoothing parameter (작을수록 정확한 DTW에 가까움)
batch_size = 512  # GPU 메모리에 따라 조절

# Soft-DTW 함수 초기화
sdtw = SoftDTW(use_cuda=torch.cuda.is_available(), gamma=gamma)

# 초기 centroid 설정 (랜덤으로 K개 선택)
torch.manual_seed(42)
perm = torch.randperm(X.size(0), device=device)
centroids = X[perm[:K]].clone()  # (K, L, 1)

print(f"\nStarting GPU-based DTW K-Means clustering...")
print(f"Clusters: {K}, Max iterations: {max_iters}")

# K-Means 반복
labels = None
for iter in range(max_iters):
    # 각 시계열과 각 centroid 간 Soft-DTW 거리 계산
    distances = []
    
    # 배치 처리로 메모리 효율성 확보
    for i in range(0, X.size(0), batch_size):
        batch = X[i:i+batch_size]  # (bs, L, 1)
        batch_size_actual = batch.size(0)
        
        # 배치 내 각 시계열과 각 centroid 간 거리 계산
        batch_distances = []
        for k in range(K):
            centroid = centroids[k:k+1]  # (1, L, 1)
            
            # Soft-DTW 거리 계산
            # batch와 centroid를 적절히 expand하여 계산
            batch_expanded = batch  # (bs, L, 1)
            centroid_expanded = centroid.expand(batch_size_actual, -1, -1)  # (bs, L, 1)
            
            # SoftDTW는 (batch, len, dim) 형태를 받음
            # 각 쌍에 대해 거리 계산
            dists = []
            for j in range(batch_size_actual):
                dist = sdtw(
                    batch_expanded[j:j+1],  # (1, L, 1)
                    centroid_expanded[j:j+1]  # (1, L, 1)
                )
                dists.append(dist)
            
            batch_dist = torch.stack(dists).squeeze()  # (bs,)
            batch_distances.append(batch_dist)
        
        # (bs, K) 형태로 변환
        batch_dist_matrix = torch.stack(batch_distances, dim=1)  # (bs, K)
        distances.append(batch_dist_matrix)
    
    # 전체 거리 행렬
    distances = torch.cat(distances, dim=0)  # (N, K)
    
    # 클러스터 할당
    labels = torch.argmin(distances, dim=1)  # (N,)
    
    # Centroid 업데이트 (각 클러스터의 평균 시계열)
    new_centroids = []
    for k in range(K):
        members = X[labels == k]  # (Nk, L, 1)
        if members.size(0) == 0:
            # 빈 클러스터: 기존 centroid 유지 또는 랜덤 재배치
            new_centroids.append(centroids[k])
        else:
            # 단순 평균 (더 정확한 방법은 Soft-DTW barycenter이지만 복잡함)
            mean_ts = members.mean(dim=0)  # (L, 1)
            new_centroids.append(mean_ts)
    
    new_centroids = torch.stack(new_centroids, dim=0)  # (K, L, 1)
    
    # 수렴 확인
    centroid_shift = torch.norm(centroids - new_centroids).item()
    centroids = new_centroids
    
    if (iter + 1) % 5 == 0:
        print(f"Iteration {iter+1}/{max_iters}, Centroid shift: {centroid_shift:.6f}")
    
    if centroid_shift < tol:
        print(f"Converged at iteration {iter+1}")
        break

# 결과를 CPU로 이동
labels = labels.cpu().numpy()
centroids = centroids.cpu().numpy()

print(f"\nClustering completed!")
print(f"Final labels shape: {labels.shape}")

# ============================================
# 결과 시각화 (기존 코드와 동일)
# ============================================
plot_count = 9
fig, axs = plt.subplots(3, 3, figsize=(25, 25))

rows = 3
columns = 3
row_i = 0
column_j = 0

for label in range(K):
    cluster_indices = np.where(labels == label)[0]
    cluster_data = [X_np[i, :, 0] for i in cluster_indices]
    
    for series in cluster_data:
        axs[row_i, column_j].plot(series, c="gray", alpha=0.4)
    
    if len(cluster_data) > 0:
        avg_series = np.average(np.vstack(cluster_data), axis=0)
        axs[row_i, column_j].plot(avg_series, c="red", linewidth=2)
    
    axs[row_i, column_j].set_title(f"Cluster {label+1} (n={len(cluster_indices)})", fontsize=30)
    
    column_j += 1
    if column_j >= columns:
        row_i += 1
        column_j = 0

plt.tight_layout()
plt.show()

# ============================================
# 클러스터 통계
# ============================================
cluster_counts = [len(labels[labels == i]) for i in range(K)]
cluster_names = [f"Cluster {i+1}" for i in range(K)]

plt.figure(figsize=(10, 10))
plt.barh(cluster_names[::-1], cluster_counts[::-1])
plt.xlabel("Number of Points")
plt.title("Cluster Distribution")
plt.show()

# ============================================
# 평가 지표 계산
# ============================================
# 주의: Soft-DTW 거리를 사용했으므로 정확한 평가를 위해서는 
# CPU에서 DTW 거리로 재계산하거나 근사치로 평가
print("\n=== Clustering Evaluation ===")
print(f"Cluster sizes: {cluster_counts}")

# ============================================
# 결과 저장
# ============================================
fancy_names_for_labels = [f"Cluster {label+1}" for label in labels]
res = pd.DataFrame(
    zip(namesofMyseries[:len(labels)], fancy_names_for_labels),
    columns=["Series", "Cluster"]
).sort_values(by="Cluster").set_index("Series")

print("\n=== Result Summary ===")
print(res.head(20))
print(f"\nTotal clustered: {len(res)}")

# 저장
# res.to_csv("kmeans_2019_w162_gpu.csv")
```
